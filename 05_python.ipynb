```python
sales = pd.read_csv("DB/sales_session4.csv")
sales
```

```python
sales["revenue"] = sales["price"] * sales["quantity"]
sales
```

```python
category_revenue = (
    sales.groupby("category")["revenue"]
    .sum()
    .sort_values(ascending=False)
)
category_revenue
```

```python
import matplotlib.pyplot as plt

category_revenue.plot.bar(
    title="Known revenue by category",
    xlabel="Category",
    ylabel="Revenue",
    color="steelblue",
    rot=0
)

plt.tight_layout()
plt.show()
```

```python
category_revenue.plot.bar(
    title="Known revenue by category",
    xlabel="Category",
    ylabel="Revenue",
    color="pink",
    rot=0
)

plt.tight_layout()
plt.show()
```

```python
sales_amount = (
    sales.groupby("category")["quantity"]
    .sum()
    .sort_values(ascending=False)
)
sales_amount.plot.bar(
    title = "The amount of units sold per category",
    xlabel = "category",
    ylabel = "units sold",
    color = "cyan",
    rot = 0
)
plt.show()
```

```python
top_5_revproducts = (
    sales.groupby("product")["revenue"]
    .sum()
    .sort_values(ascending=False)
).head(5)
top_5_revproducts.plot.bar(
    title = "Top 5 products by revenue",
    xlabel = "Product",
    ylabel = "revenue per product",
    color ="black",
    rot = 0
)
plt.gca().set_axisbelow(True)
plt.grid(axis="y", alpha=0.55)
plt.show()
```

```python
avg_price_pcat = (
    sales.groupby("category")["price"]
    .mean()
    .sort_values(ascending=False)
)
avg_price_pcat.plot.bar(
    title = "Average price by category",
    xlabel = "category",
    ylabel = "average price",
    color = "purple",
    rot = 0
)
plt.gca().set_axisbelow(True)
plt.grid(axis="y", alpha=0.55)
plt.show()
```

```python
sales.plot.scatter(
    x="price",
    y="quantity",
    color = "black",
    title = "plot price and quantity")
```

```python
import seaborn as sns

sns.regplot(data=sales, x="price", y="quantity", ci=None)

plt.title("Price vs quantity sold")
plt.show()
sales["price"].corr(sales["quantity"])
```
