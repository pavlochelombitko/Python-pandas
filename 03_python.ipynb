```python
def classify_revenue(value):
    if value >= 35:
        return "High"
    elif value >= 20:
        return "Medium"
    else:
        return "Low"
def classify_order(value):
    if value >= 6:
        return "Large"
    else:
        return "Small"
sales["category"] = sales["revenue"].apply(classify_revenue)
sales["order_size"] = sales["quantity"].apply(classify_order)
sales
```
