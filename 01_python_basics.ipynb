```python
sales = [350, 420, 280, 510, 390]
target = 400

for sale in sales:
    if sale > target:
        print(sale, "is above target")
    else:
        print(sale, "is not above target")
```

```python
profits = [120, -30, 75, -15, 200, 0, 95]
target = 0
profitable_count = 0
loss_count = 0
break_even_count = 0
for profit in profits:
    if profit > target:
        print(profit, "is profitable")
        profitable_count += 1
    elif profit < target:
        print(profit, "is a loss")
        loss_count += 1
    else:
        print(profit, "is break-even")
        break_even_count += 1

print("Profitable days:", profitable_count)
print("Loss days:", loss_count)
print("Break-even days:", break_even_count)
```

```python
# we use def to make a function, that can save us time
def calculate_average(values):
    total = sum(values)
    average = total / len(values)
    return average

sales_week_1 = [350, 420, 280, 510, 390]
sales_week_2 = [440, 380, 525, 475, 600]

print("Average sales for week 1:", calculate_average(sales_week_1))
print("Average sales for week 2:", calculate_average(sales_week_2))
```

```python
def percentage_change(old_value, new_value):
    change = new_value - old_value
    percentage = (change / old_value) * 100
    return percentage

last_year_sales = 40000
this_year_sales = 46000
print("Percentage sale change", percentage_change(last_year_sales, this_year_sales))
```

```python
def percentage_change(old_value, new_value):
    change = new_value - old_value
    percentage = (change / old_value) * 100
    return percentage

first_value = 500
second_value = 425

print("Percentage change:", percentage_change(first_value, second_value), "%")
```

```python
import pandas as pd

data = {
    "product": ["Laptop", "Monitor", "Keyboard", "Mouse"],
    "price": [850, 240, 75, 35],
    "quantity": [3, 5, 10, 20]
}

df = pd.DataFrame(data)
df
```

```python
df["revenue"] = df["price"] * df["quantity"]
df
```

```python
print("Total revenue:", df["revenue"].sum())
print("Average price:", df["price"].mean())
print("Most expensive:", df["price"].max())
```

```python
expensive_product = df[df["price"] > 100]
expensive_product
```

```python
df["price_with_tax"] = df["price"] * 1.20
```

```python
df
```

```python
highest_revenue_product = df.loc[df["revenue"].idxmax()]
print(highest_revenue_product)
```

```python
df["revenue_with_tax"] = df["price_with_tax"] * df["quantity"]
df
```

```python
df.loc[df["revenue_with_tax"].idxmax()]
```
