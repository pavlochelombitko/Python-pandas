```python
daily_sales = [450, 520, 390, 610, 480, 700, 550]

total_sales = sum(daily_sales)
print("Total sales:", total_sales)
average_sales = total_sales / len(daily_sales)
print("Average daily sales:", round(average_sales, 2))
highest_sale = max(daily_sales)
print("Highest daily sale:", highest_sale)
lowest_sale = min(daily_sales)
print("Lowest daily sale:", lowest_sale)
first_sale = daily_sales[0]
print("First daily sale:", first_sale)
last_sale = daily_sales[-1]
print("Last daily sale:", last_sale)
diff_high_low = highest_sale - lowest_sale
print("Difference between highest and lowest daily sales:", diff_high_low)
```

```python
# task 3

daily_sales = [450, 520, 390, 610, 480, 700, 550]
target = 500
above_target = 0
below_target = 0
on_target = 0
for sale in daily_sales:
    if sale > target:
        print(sale, "is above the target of", target)
        above_target += 1
    elif sale == target:
        print(sale, "is equal to the target of", target)
        on_target += 1
    else:
        print(sale, "is below the target of", target)
        below_target += 1

print("Number of sales above target:", above_target)
print("Number of sales below target:", below_target)
print("Number of sales on target:", on_target)
```

```python
# part 4. funcions. def
#percentage change function

def percentage_change(new_value, old_value):
    if old_value == 0:
        return "Old value cannot be zero for percentage change calculation."
    change = new_value - old_value
    percent_change = (change / old_value) * 100
    return percent_change

first_old = 12000
first_new = 15000
second_old = 8000
second_new = 6800
third_old = 500
third_new = 500
print("Percentage change from", first_old, "to", first_new, "is:", percentage_change(first_new, first_old), "%")
print("Percentage change from", second_old, "to", second_new, "is:", percentage_change(second_new, second_old), "%")
print("Percentage change from", third_old, "to", third_new, "is:", percentage_change(third_new, third_old), "%")
```
