```python
import pandas as pd
```

```python
sales = pd.DataFrame({
    "product": ["Backpack", "Pen", "Notebook", "Lamp"],
    "category": ["Accessories", "Stationery", "Stationery", "Home"],
    "price": [40, 2, 8, 25],
    "quantity": [2, 30, 10, 3]
}, index=[10, 20, 30, 40])

sales
```

```python
sales["revenue"] = sales["price"] * sales["quantity"]
sales
```

```python
sales.loc[sales["price"].idxmax(), "product"]
```

```python
sales.loc[sales["revenue"].idxmax(), "product"]
```

```python
sales.loc[sales["revenue"] == sales["revenue"].max(), "product"]
```

```python
stationery = sales.loc[sales["category"] == "Stationery"]
stationery
```

```python
stationery["revenue"].sum()
```

```python
stationery.loc[stationery["price"].idxmax(), "product"]
```
